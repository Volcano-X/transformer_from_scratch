```python
# Indices to be one-hot encoded
indices = torch.tensor([0, 2, 1, 3])

# Number of classes
num_classes = 5

# Create a zero tensor for one-hot encoding
one_hot = torch.zeros(indices.size(0), num_classes)
```

```python
indices.unsqueeze(1)
```

```python
# One-hot encode using scatter_
one_hot.scatter_(1, indices.unsqueeze(1), 1)

print(one_hot)
```
